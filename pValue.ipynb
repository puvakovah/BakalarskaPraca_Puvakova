```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

mu = 0       # stredná hodnota pod nulovou hypotézou
sigma = 1    # štandardná odchýlka
x_obs = 1.5

x = np.linspace(-4, 4, 500)
y = stats.norm.pdf(x, mu, sigma)

p_value = 1 - stats.norm.cdf(x_obs, mu, sigma)

plt.figure(figsize=(10, 6))
plt.plot(x, y,
         label='hustota pravdepodobnosti\n'
               'testovacej štatistiky za platnosti $H_0$',
         color='blue', lw=2)

# Vyfarbenie plochy, ktorá predstavuje p-hodnotu
x_fill = np.linspace(x_obs, 4, 100)
y_fill = stats.norm.pdf(x_fill, mu, sigma)
plt.fill_between(x_fill, y_fill, color='red', alpha=0.5, label=f'p-hodnota')

# Označenie pozorovanej hodnoty
plt.axvline(x_obs, color='black', linestyle='--', label=f'pozorovaná hodnota ')

plt.xlabel('Hodnota štatistiky')
plt.ylabel('Hustota pravdepodobnosti')
plt.legend()
plt.grid(False)  
plt.savefig("p_value", dpi=300)
plt.show()
```

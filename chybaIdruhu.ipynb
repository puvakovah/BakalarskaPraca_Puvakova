```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# nastavenie parametrov
alpha = 0.05
# rozsah hodnot
m_values = np.linspace(1, 100, 200)
m_points = [1, 5, 10, 20, 50, 100]

# pravdepodobnost ze aspon jeden test bude signifikantny
theoretical_probs = 1 - (1 - alpha)**m_values
point_probs = [1 - (1 - alpha)**m for m in m_points]

df = pd.DataFrame({
    'm': m_points,
    'P': [round(p, 4) for p in point_probs]
})
df.to_csv('tabulka_inflacia.csv', index=False)

# vizualizacia
plt.figure(figsize=(10, 6))
plt.plot(m_values, theoretical_probs, 
         label=r'$P(P_{min} \leq \alpha) = 1 - (1 - \alpha)^m$', 
         color='tab:blue', linewidth=2)
plt.scatter(m_points, point_probs, color='red', s=50, zorder=5)

plt.title("Analytické vyjadrenie inflácie chyby I. druhu", fontsize=14)
plt.xlabel("Počet testov (m)", fontsize=12)
plt.ylabel("Pravdepodobnosť aspoň jednej chyby", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(loc='upper left')
plt.ylim(-0.05, 1.05)

plt.show()
```
